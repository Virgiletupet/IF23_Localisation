```python
import tkinter as tk
from tkinter import ttk, messagebox
import threading
import time
import pandas as pd
import numpy as np
from pathlib import Path
import joblib
import json
import pywifi
from pywifi import const
from collections import deque

# ============================================================
# CONFIGURATION
# ============================================================

MODEL_DIR = Path("../models/legacy")
CLEANED_DATA_DIR = Path("../data/cleaned")

SCAN_INTERVAL_MS = 500  # Scan toutes les 500ms
SMOOTHING_WINDOW = 5    # Moyenne glissante sur les 5 dernières prédictions

# ============================================================
# SCANNER WIFI
# ============================================================

class WiFiScanner:
    """Scanner WiFi temps réel"""
    
    def __init__(self):
        wifi = pywifi.PyWiFi()
        if len(wifi.interfaces()) == 0:
            raise Exception("❌ Aucune interface WiFi détectée")
        self.iface = wifi.interfaces()[0]
        print(f"✅ Interface WiFi: {self.iface.name()}")
    
    def scan(self):
        """
        Scan les réseaux WiFi et retourne {bssid: rssi}
        
        Returns:
            dict: {bssid: rssi}
        """
        try:
            self.iface.scan()
            time.sleep(0.3)  # Attendre la fin du scan
            
            results = self.iface.scan_results()
            rssi_dict = {}
            
            for network in results:
                # Normaliser le BSSID (minuscules, sans ':' final)
                bssid = network.bssid.lower().strip().rstrip(':')
                rssi = network.signal
                
                # Garder le meilleur signal si doublon
                if bssid not in rssi_dict or rssi > rssi_dict[bssid]:
                    rssi_dict[bssid] = rssi
            
            return rssi_dict
            
        except Exception as e:
            print(f"⚠️ Erreur scan: {e}")
            return {}

# ============================================================
# PRÉDICTEUR
# ============================================================

class RoomPredictor:
    """Prédicteur de salle basé sur les signatures WiFi"""
    
    def __init__(self):
        self.model = None
        self.label_encoder = None
        self.feature_names = []  # Liste des BSSID attendus (dans l'ordre)
        self.class_names = []
        self.metadata = {}
        self.is_loaded = False
    
    def load(self):
        """Charge le modèle et les métadonnées"""
        
        model_file = MODEL_DIR / "room_classifier.pkl"
        encoder_file = MODEL_DIR / "label_encoder.pkl"
        metadata_file = MODEL_DIR / "model_metadata.json"
        
        # Vérifier l'existence des fichiers
        if not model_file.exists():
            raise FileNotFoundError(f"Modèle introuvable: {model_file}")
        
        if not encoder_file.exists():
            raise FileNotFoundError(f"Encodeur introuvable: {encoder_file}")
        
        if not metadata_file.exists():
            raise FileNotFoundError(f"Métadonnées introuvables: {metadata_file}")
        
        # Charger le modèle
        self.model = joblib.load(model_file)
        print(f"✅ Modèle chargé: {model_file.name}")
        
        # Charger l'encodeur
        self.label_encoder = joblib.load(encoder_file)
        print(f"✅ Encodeur chargé: {encoder_file.name}")
        
        # Charger les métadonnées
        with open(metadata_file, 'r') as f:
            self.metadata = json.load(f)
        
        self.feature_names = self.metadata['feature_names']
        self.class_names = self.metadata['class_names']
        
        print(f"✅ Métadonnées chargées:")
        print(f"   📍 {len(self.class_names)} salles")
        print(f"   📡 {len(self.feature_names)} BSSID")
        print(f"   🎯 Test Accuracy: {self.metadata['test_score']:.2%}")
        
        self.is_loaded = True
    
    def create_feature_vector(self, rssi_dict):
        """
        Crée un vecteur de features à partir du scan WiFi
        
        Args:
            rssi_dict: {bssid: rssi} du scan actuel
        
        Returns:
            np.array: Vecteur de features dans l'ordre attendu par le modèle
        """
        
        feature_vector = []
        
        for bssid in self.feature_names:
            if bssid in rssi_dict:
                feature_vector.append(rssi_dict[bssid])
            else:
                # BSSID absent = signal très faible
                feature_vector.append(-100)
        
        return np.array(feature_vector).reshape(1, -1)
    
    def predict(self, rssi_dict):
        """
        Prédit la salle à partir d'un scan WiFi
        
        Args:
            rssi_dict: {bssid: rssi} du scan actuel
        
        Returns:
            dict: {
                'room': str,
                'confidence': float,
                'probabilities': {room: proba}
            }
        """
        
        if not self.is_loaded:
            raise ValueError("Modèle non chargé !")
        
        # Créer le vecteur de features
        X = self.create_feature_vector(rssi_dict)
        
        # Prédire avec probabilités
        probas = self.model.predict_proba(X)[0]
        predicted_class = self.model.predict(X)[0]
        
        # Convertir en nom de salle
        predicted_room = self.label_encoder.inverse_transform([predicted_class])[0]
        confidence = probas[predicted_class]
        
        # Créer le dictionnaire {room: proba}
        probabilities = {}
        for idx, room in enumerate(self.class_names):
            probabilities[room] = probas[idx]
        
        return {
            'room': predicted_room,
            'confidence': confidence,
            'probabilities': probabilities
        }
    
    def get_matched_bssids(self, rssi_dict):
        """
        Retourne les BSSID du scan qui matchent avec le modèle
        
        Args:
            rssi_dict: {bssid: rssi} du scan
        
        Returns:
            dict: {bssid: rssi} filtré
        """
        matched = {
            bssid: rssi 
            for bssid, rssi in rssi_dict.items() 
            if bssid in self.feature_names
        }
        return matched
    
    def get_coverage(self, rssi_dict):
        """
        Calcule le pourcentage de BSSID détectés
        
        Args:
            rssi_dict: {bssid: rssi} du scan
        
        Returns:
            float: Pourcentage de couverture (0-1)
        """
        matched = self.get_matched_bssids(rssi_dict)
        return len(matched) / len(self.feature_names)

# ============================================================
# INTERFACE GRAPHIQUE
# ============================================================

class RoomLocalizerGUI:
    """Interface graphique de localisation en temps réel"""
    
    def __init__(self):
        self.root = tk.Tk()
        self.root.title("🎯 WiFi Room Localizer - Temps Réel")
        self.root.geometry("1000x800")
        self.root.configure(bg='#1e1e1e')
        
        # Composants
        self.scanner = None
        self.predictor = RoomPredictor()
        
        # État du scan
        self.is_scanning = False
        self.scan_thread = None
        
        # Historique pour lissage
        self.prediction_history = deque(maxlen=SMOOTHING_WINDOW)
        
        # UI
        self.setup_ui()
        
        # Charger le modèle automatiquement
        self.root.after(100, self.auto_load_model)
    
    def setup_ui(self):
        """Crée l'interface"""
        
        # Style
        style = ttk.Style()
        style.theme_use('clam')
        
        # ==== HEADER ====
        header_frame = tk.Frame(self.root, bg='#1e1e1e', pady=20)
        header_frame.pack(fill='x')
        
        title = tk.Label(
            header_frame,
            text="🎯 WiFi Room Localizer",
            bg='#1e1e1e',
            fg='white',
            font=('Arial', 20, 'bold')
        )
        title.pack()
        
        subtitle = tk.Label(
            header_frame,
            text="Localisation en temps réel par signature WiFi",
            bg='#1e1e1e',
            fg='#888',
            font=('Arial', 10)
        )
        subtitle.pack(pady=5)
        
        self.status_label = tk.Label(
            header_frame,
            text="⚠️ Modèle non chargé",
            bg='#1e1e1e',
            fg='#ff8800',
            font=('Arial', 11, 'bold')
        )
        self.status_label.pack(pady=5)
        
        # ==== CONTROLS ====
        control_frame = tk.Frame(self.root, bg='#1e1e1e', pady=15)
        control_frame.pack(fill='x')
        
        self.load_btn = tk.Button(
            control_frame,
            text="📂 Charger Modèle",
            command=self.load_model,
            bg='#00aa00',
            fg='white',
            font=('Arial', 11, 'bold'),
            padx=20,
            pady=8,
            relief='flat',
            cursor='hand2'
        )
        self.load_btn.pack(side='left', padx=10)
        
        self.scan_btn = tk.Button(
            control_frame,
            text="▶️ Démarrer Scan",
            command=self.toggle_scanning,
            bg='#0066cc',
            fg='white',
            font=('Arial', 11, 'bold'),
            padx=20,
            pady=8,
            relief='flat',
            cursor='hand2',
            state='disabled'
        )
        self.scan_btn.pack(side='left', padx=10)
        
        self.reset_btn = tk.Button(
            control_frame,
            text="🔄 Réinitialiser",
            command=self.reset,
            bg='#666',
            fg='white',
            font=('Arial', 11, 'bold'),
            padx=20,
            pady=8,
            relief='flat',
            cursor='hand2'
        )
        self.reset_btn.pack(side='left', padx=10)
        
        # ==== CURRENT LOCATION ====
        location_frame = tk.LabelFrame(
            self.root,
            text="📍 Localisation Actuelle",
            bg='#2a2a2a',
            fg='white',
            font=('Arial', 13, 'bold'),
            padx=20,
            pady=20
        )
        location_frame.pack(fill='x', padx=20, pady=15)
        
        self.current_room_label = tk.Label(
            location_frame,
            text="---",
            bg='#2a2a2a',
            fg='#00ff00',
            font=('Arial', 36, 'bold')
        )
        self.current_room_label.pack()
        
        self.confidence_label = tk.Label(
            location_frame,
            text="Confiance: ---%",
            bg='#2a2a2a',
            fg='#aaa',
            font=('Arial', 14)
        )
        self.confidence_label.pack(pady=5)
        
        # ==== PROBABILITIES ====
        proba_frame = tk.LabelFrame(
            self.root,
            text="📊 Probabilités par Salle",
            bg='#2a2a2a',
            fg='white',
            font=('Arial', 13, 'bold'),
            padx=15,
            pady=15
        )
        proba_frame.pack(fill='both', expand=True, padx=20, pady=15)
        
        # Scrollable canvas
        canvas = tk.Canvas(proba_frame, bg='#2a2a2a', highlightthickness=0)
        scrollbar = ttk.Scrollbar(proba_frame, orient='vertical', command=canvas.yview)
        
        self.proba_content = tk.Frame(canvas, bg='#2a2a2a')
        
        canvas.create_window((0, 0), window=self.proba_content, anchor='nw')
        canvas.configure(yscrollcommand=scrollbar.set)
        
        canvas.pack(side='left', fill='both', expand=True)
        scrollbar.pack(side='right', fill='y')
        
        self.proba_content.bind(
            '<Configure>',
            lambda e: canvas.configure(scrollregion=canvas.bbox('all'))
        )
        
        self.proba_bars = {}
        
        # ==== NETWORK INFO ====
        info_frame = tk.Frame(self.root, bg='#1e1e1e', pady=10)
        info_frame.pack(fill='x', padx=20, pady=10)
        
        self.network_info_label = tk.Label(
            info_frame,
            text="📡 Réseaux: 0 détectés | 0 utilisés (0%)",
            bg='#1e1e1e',
            fg='#888',
            font=('Arial', 10)
        )
        self.network_info_label.pack()
    
    def auto_load_model(self):
        """Charge automatiquement le modèle au démarrage"""
        try:
            self.load_model()
        except:
            pass
    
    def load_model(self):
        """Charge le modèle"""
        try:
            self.status_label.config(text="⏳ Chargement du modèle...")
            self.root.update()
            
            # Charger le modèle
            self.predictor.load()
            
            # Initialiser le scanner
            self.scanner = WiFiScanner()
            
            self.status_label.config(
                text=f"✅ Modèle chargé - {len(self.predictor.class_names)} salles - "
                f"Accuracy: {self.predictor.metadata['test_score']:.1%}",
                fg='#00ff00'
            )
            
            self.scan_btn.config(state='normal')
            self.load_btn.config(state='disabled')
            
            messagebox.showinfo(
                "Succès",
                f"✅ Modèle chargé avec succès !\n\n"
                f"📍 Salles: {len(self.predictor.class_names)}\n"
                f"📡 BSSID: {len(self.predictor.feature_names)}\n"
                f"🎯 Test Accuracy: {self.predictor.metadata['test_score']:.2%}"
            )
            
        except FileNotFoundError as e:
            self.status_label.config(
                text="❌ Modèle introuvable",
                fg='#ff4444'
            )
            messagebox.showerror(
                "Erreur",
                f"Fichier introuvable:\n{e}\n\n"
                "Entraînez d'abord le modèle (notebook 2)."
            )
        except Exception as e:
            self.status_label.config(
                text="❌ Erreur de chargement",
                fg='#ff4444'
            )
            messagebox.showerror("Erreur", f"Erreur:\n{e}")
    
    def toggle_scanning(self):
        """Démarre/arrête le scan"""
        if not self.is_scanning:
            self.start_scanning()
        else:
            self.stop_scanning()
    
    def start_scanning(self):
        """Démarre le scan"""
        self.is_scanning = True
        self.scan_btn.config(
            text="⏸️ Arrêter Scan",
            bg='#cc0000'
        )
        self.load_btn.config(state='disabled')
        self.reset_btn.config(state='disabled')
        
        # Créer les barres
        self.create_proba_bars(self.predictor.class_names)
        
        # Réinitialiser l'historique
        self.prediction_history.clear()
        
        # Lancer le thread de scan
        self.scan_thread = threading.Thread(target=self.scan_loop, daemon=True)
        self.scan_thread.start()
    
    def stop_scanning(self):
        """Arrête le scan"""
        self.is_scanning = False
        self.scan_btn.config(
            text="▶️ Démarrer Scan",
            bg='#0066cc'
        )
        self.reset_btn.config(state='normal')
    
    def reset(self):
        """Réinitialise l'affichage"""
        self.current_room_label.config(text="---", fg='#00ff00')
        self.confidence_label.config(text="Confiance: ---%")
        self.network_info_label.config(text="📡 Réseaux: 0 détectés | 0 utilisés (0%)")
        
        for widget in self.proba_content.winfo_children():
            widget.destroy()
        
        self.proba_bars = {}
        self.prediction_history.clear()
    
    def create_proba_bars(self, rooms):
        """Crée les barres de probabilité"""
        
        # Nettoyer
        for widget in self.proba_content.winfo_children():
            widget.destroy()
        
        self.proba_bars = {}
        
        for room in sorted(rooms):
            room_frame = tk.Frame(self.proba_content, bg='#2a2a2a', pady=5)
            room_frame.pack(fill='x', padx=10, pady=3)
            
            # Nom de la salle
            name_label = tk.Label(
                room_frame,
                text=room,
                bg='#2a2a2a',
                fg='white',
                font=('Arial', 11, 'bold'),
                width=18,
                anchor='w'
            )
            name_label.pack(side='left', padx=8)
            
            # Barre de progression
            bar_canvas = tk.Canvas(
                room_frame,
                width=600,
                height=24,
                bg='#3a3a3a',
                highlightthickness=0
            )
            bar_canvas.pack(side='left', padx=8)
            
            bar_rect = bar_canvas.create_rectangle(
                0, 0, 0, 24,
                fill='#0066cc',
                outline=''
            )
            
            # Label pourcentage
            pct_label = tk.Label(
                room_frame,
                text="0.0%",
                bg='#2a2a2a',
                fg='#aaa',
                font=('Arial', 11),
                width=8,
                anchor='e'
            )
            pct_label.pack(side='left', padx=8)
            
            self.proba_bars[room] = (bar_canvas, bar_rect, pct_label)
    
    def scan_loop(self):
        """Boucle principale de scan"""
        while self.is_scanning:
            try:
                # 1. Scanner les réseaux WiFi
                rssi_dict = self.scanner.scan()
                
                if not rssi_dict:
                    print("⚠️ Aucun réseau détecté")
                    time.sleep(SCAN_INTERVAL_MS / 1000.0)
                    continue
                
                # 2. Prédire la salle
                result = self.predictor.predict(rssi_dict)
                
                # 3. Ajouter à l'historique pour lissage
                self.prediction_history.append(result)
                
                # 4. Calculer la prédiction lissée (moyenne des probas)
                smoothed_probas = {}
                for room in self.predictor.class_names:
                    probas = [r['probabilities'][room] for r in self.prediction_history]
                    smoothed_probas[room] = np.mean(probas)
                
                best_room = max(smoothed_probas, key=smoothed_probas.get)
                best_confidence = smoothed_probas[best_room]
                
                # 5. Calculer les stats réseaux
                matched_bssids = self.predictor.get_matched_bssids(rssi_dict)
                coverage = self.predictor.get_coverage(rssi_dict)
                
                # 6. Mettre à jour l'interface (dans le thread principal)
                self.root.after(0, lambda: self.update_display(
                    best_room,
                    best_confidence,
                    smoothed_probas,
                    len(rssi_dict),
                    len(matched_bssids),
                    coverage
                ))
                
                # 7. Attendre avant le prochain scan
                time.sleep(SCAN_INTERVAL_MS / 1000.0)
                
            except Exception as e:
                print(f"❌ Erreur dans scan_loop: {e}")
                import traceback
                traceback.print_exc()
                time.sleep(1)
    
    def update_display(self, best_room, best_confidence, probas, 
                      total_networks, matched_networks, coverage):
        """Met à jour l'affichage"""
        
        # 1. Localisation actuelle
        self.current_room_label.config(text=best_room)
        self.confidence_label.config(text=f"Confiance: {best_confidence:.1%}")
        
        # 2. Couleur selon confiance
        if best_confidence > 0.7:
            color = '#00ff00'  # Vert
        elif best_confidence > 0.4:
            color = '#ffaa00'  # Orange
        else:
            color = '#ff4444'  # Rouge
        
        self.current_room_label.config(fg=color)
        
        # 3. Barres de probabilité
        for room, proba in probas.items():
            if room in self.proba_bars:
                canvas, rect, label = self.proba_bars[room]
                
                # Largeur de la barre
                bar_width = int(600 * proba)
                canvas.coords(rect, 0, 0, bar_width, 24)
                
                # Couleur
                if room == best_room:
                    bar_color = color
                else:
                    bar_color = '#0066cc'
                
                canvas.itemconfig(rect, fill=bar_color)
                
                # Label
                label.config(text=f"{proba:.1%}")
        
        # 4. Info réseaux
        self.network_info_label.config(
            text=f"📡 Réseaux: {total_networks} détectés | "
                 f"{matched_networks} utilisés ({coverage:.0%})"
        )
    
    def run(self):
        """Lance l'application"""
        self.root.mainloop()

# ============================================================
# LANCEMENT
# ============================================================

if __name__ == "__main__":
    print("=" * 60)
    print("🎯 WiFi Room Localizer - Interface Graphique")
    print("=" * 60)
    print()
    
    try:
        app = RoomLocalizerGUI()
        app.run()
    except Exception as e:
        print(f"❌ Erreur critique: {e}")
        import traceback
        traceback.print_exc()
```
